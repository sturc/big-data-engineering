# Sales Data Transformation with Pipelining

```python
import os
import sys
sys.path.append("..")
# Spark libs
from pyspark.sql.session import SparkSession
from pyspark.ml.feature import OneHotEncoder, StringIndexer, QuantileDiscretizer, VectorAssembler , MinMaxScaler, StandardScaler
from pyspark.ml import Pipeline
# helpers
from helpers.data_prep_and_print import print_df
```

### Select the Imput File

```python
inputFile = f"s3a://{os.environ['LAB_BUCKET']}/data/sales.csv"
#inputFile = "../data/sales.csv"
```

### Spark Session Creation

```python
spark = (SparkSession
       .builder
       .appName("Sales Data Transformation with Pipelining")
       .getOrCreate())
```

### Create Dataframe from csv File

```python
df = spark.read.option("header", "true") \
       .option("inferSchema", "true") \
       .option("delimiter", ",") \
       .csv(inputFile)
print(df.printSchema())
```

## Transformations via Pipeline 

```python
feature_cols = ["training level","work experience","salary","sales_bucket_quantile","division_num","education_cat_vector"]
assembler =  VectorAssembler(outputCol="features", inputCols=list(feature_cols))


pipeline = Pipeline(stages=[
    StringIndexer(inputCol="division", outputCol="division_num"),
    StringIndexer(inputCol="level of education", outputCol="education_num"),
    OneHotEncoder(inputCol="education_num", outputCol="education_cat_vector", dropLast=False),
    QuantileDiscretizer(numBuckets=10, inputCol="sales", outputCol="sales_bucket_quantile"),
    assembler,
    MinMaxScaler(inputCol="features", outputCol="norm_features"),
    StandardScaler(inputCol="features", outputCol="scaledFeatures",withStd=True, withMean=True)
])
pipeline_model = pipeline.fit(df)
df_encoded = pipeline_model.transform(df)
print_df (df_encoded,10)
```

```python
spark.stop()
```
