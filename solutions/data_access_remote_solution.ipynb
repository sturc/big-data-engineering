# Remote Data Access Exercise

```python
import os
import sys
sys.path.append("..")
from pyspark.sql import DataFrameReader
from pyspark.sql import SparkSession
from helpers.path_translation import translate_to_file_string
from helpers.data_prep_and_print import print_df
```

### Create spark session

```python
#create a SparkSession
spark = (SparkSession
       .builder
     #  .remote("sc://spark-connect:15002")
       .appName("Data Access")
       .getOrCreate())
```

### Read CSV File

```python
# create a DataFrame using an ifered Schema 
inputFile = f"s3a://{os.environ['LAB_BUCKET']}/data/churn.csv"
df = spark.read.option("header", "true") \
       .option("inferSchema", "true") \
       .option("delimiter", ";") \
       .csv(inputFile)   
print_df(df,10)
```

### Read Json file

```python
# create a DataFrame from Json
inputFile = f"s3a://{os.environ['LAB_BUCKET']}/data/census_2010.json"
df = spark.read.json(inputFile)   
print_df(df,10)
```

### Read text file

```python
from pyspark.sql import functions as F

inputFile = f"s3a://{os.environ['LAB_BUCKET']}/data/ghEmployees.txt"
input = spark.read.text(inputFile)
counts = (
    input
    .select(F.explode(F.split(F.col("value"), ";")).alias("word"))
    .groupBy("word")
    .count()
    .orderBy(F.desc("count"))
)
print_df(counts, 10)
```

### Read Parquet Files

```python
inputFile = f"s3a://{os.environ['LAB_BUCKET']}/data/census_2010.parquet"
df = spark.read.parquet(inputFile)   
print_df(df,10)
```
